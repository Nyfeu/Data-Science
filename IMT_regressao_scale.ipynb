# Q1.

Considere somente os dados numéricos, sem transformações. Qual o R2 obtido da regressão linear para estimativa do preço dos imóveis?

```python
import statsmodels.formula.api as smf

# Criando o modelo de regressão linear usando ols (Ordinary Least Squares)
model = smf.ols(formula='preco ~ areaM2 + suites + dormitorios + banheiros + vagas', data=df).fit()

# Exibindo o sumário estatístico do modelo
print(model.summary())

# Acessando diretamente o valor de R2
print(f'\nO R2 obtido é: {model.rsquared}')
```

# Q2.

Considere somente os dados numéricos, sem transformações. Quais atributos preditores são significativos?

```python
# Acessando os p-values do modelo treinado na célula anterior
p_values = model.pvalues

# Filtrando apenas os atributos com p-value menor que 0.05 (nível de significância de 5%)
atributos_significativos = p_values[p_values < 0.05]

print("Atributos preditores significativos (p-value < 0.05):")
print(atributos_significativos)

# Podemos listar os nomes (removendo o Intercept)
nomes_significativos = [attr for attr in atributos_significativos.index if attr != 'Intercept']
print(f"\nAs variáveis significativas são: {', '.join(nomes_significativos)}")
```

# Q3.

Considere somente os dados numéricos, sem transformações. Qual o R2 obtido somente com os coeficientes significativos?

```python
# Criando o modelo de regressão linear usando ols (Ordinary Least Squares)
model = smf.ols(formula='preco ~ areaM2 + suites + dormitorios + vagas - 1', data=df).fit()

# Exibindo o sumário estatístico do modelo
print(model.summary())

# Acessando diretamente o valor de R2
print(f'\nO R2 obtido é: {model.rsquared}')
```

# Q4.

Considere somente os dados numéricos, sem transformações. Considerando somente os coeficientes significativos da regressão linear, qual o preço estimado de um imóvel de 134 m2, uma suite, 4 dormitórios e 2 vagas de garagem?


```python
import pandas as pd

# Criando um DataFrame para a previsão com os nomes das colunas corretos
# A ordem deve corresponder à ordem das variáveis no modelo: areaM2, suites, dormitorios, vagas
dados_para_prever = pd.DataFrame([[134, 1, 4, 2]], columns=['areaM2', 'suites', 'dormitorios', 'vagas'])

preco = model.predict(dados_para_prever)
print(f'O preço estimado é: R$ {preco[0]:.2f}')
```

# Q6.

Normalize os dados, incluindo o preço, apenas dividindo pelo valor máximo de cada atributo e refaça a regressão linear somente os preditores significativos. Qual a estimativa do preço agora?

```python
# Identificando as colunas numéricas para normalização (excluindo 'bairro')
colunas_numericas = ['areaM2', 'suites', 'dormitorios', 'banheiros', 'vagas', 'preco']

# Calculando o valor máximo para cada coluna numérica
max_values = df[colunas_numericas].max()

# Criando um DataFrame normalizado
df_normalized = df[colunas_numericas] / max_values

# Reconstruindo o modelo de regressão linear com os preditores significativos e dados normalizados
# Usando os nomes significativos identificados anteriormente: 'areaM2', 'suites', 'dormitorios', 'vagas'
model_normalized = smf.ols(formula='preco ~ areaM2 + suites + dormitorios + vagas - 1', data=df_normalized).fit()

print("\nSumário do Modelo de Regressão Linear com Dados Normalizados:")
print(model_normalized.summary())

print(f'\nO R2 obtido do modelo normalizado é: {model_normalized.rsquared}')

# Normalizando os dados de entrada para previsão (134 m2, 1 suite, 4 dormitórios, 2 vagas)
# A ordem das colunas no DataFrame de previsão deve ser a mesma do modelo
input_data_normalized = pd.DataFrame([[
    134 / max_values['areaM2'],
    1 / max_values['suites'],
    4 / max_values['dormitorios'],
    2 / max_values['vagas']
]], columns=['areaM2', 'suites', 'dormitorios', 'vagas'])

# Fazendo a previsão do preço normalizado
preco_normalizado_previsto = model_normalized.predict(input_data_normalized)

# Denormalizando o preço previsto
preco_denormalizado_previsto = preco_normalizado_previsto[0] * max_values['preco']

print(f'\nPreço estimado para o imóvel (após normalização e denormalização): R$ {preco_denormalizado_previsto:.2f}')

# --- Predição sem normalização dos dados de entrada ---

# Criando um DataFrame para a previsão sem normalização (mesmos valores)
input_data_unnormalized = pd.DataFrame([[134, 1, 4, 2]], columns=['areaM2', 'suites', 'dormitorios', 'vagas'])

# Fazendo a previsão com o modelo original (não normalizado), que foi definido como 'model'
preco_unnormalized_previsto = model.predict(input_data_unnormalized)

print(f'\nPreço estimado para o imóvel (sem normalização dos dados de entrada): R$ {preco_unnormalized_previsto[0]:.2f}')

print("\n" + "-"*60)
print("Explicação: Por que os valores deram iguais?")
print("Na Regressão Linear (OLS), dividir variáveis (incluindo o target) por uma constante (como o MAX)")
print("apenas redimensiona os coeficientes (pesos) proporcionalmente. O relacionamento matemático entre")
print("os preditores e o preço é mantido de forma idêntica.")
print("Ao desnormalizar o resultado (multiplicando pelo MAX do preço), voltamos exatamente à mesma equação")
print("do modelo não-normalizado. Portanto, para a regressão linear OLS, é esperado que a previsão")
print("seja matematicamente idêntica!")
```
